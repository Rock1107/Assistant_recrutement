```python
candidats["poste_cible"].value_counts()
```

```python
candidats["poste_cible"].value_counts().plot(
    kind="bar",
    figsize=(8,5)
)

plt.title("Repartition des candidats selon leur poste cible")
plt.xlabel("poste_cible")
plt.ylabel("Nombre de candidats")
plt.xticks(rotation=45)
plt.show
```

```python
candidats["diplome"].value_counts()
```

```python
candidats["diplome"].value_counts().plot(
    kind="bar",
    figsize=(10,5)
)

plt.title("Repartition des candidats selon le diplome")
plt.xlabel("diplome")
plt.ylabel("Nombre de candidats")
plt.xticks(rotation=45)
plt.show
```

```python
candidats["experience"].value_counts()
```

```python
candidats["experience_annees"] = (
    candidats["experience"]
    .str.extract(r"(\d+)")
    .astype(int)
)
```

```python
candidats[["experience", "experience_annees"]].head()
```

```python
candidats["experience_annees"].describe()
```

```python
candidats["experience_annees"].value_counts().sort_index().plot(
    kind="bar",
    figsize=(8, 5)
)

plt.title("Distribution de l'expérience professionnelle des candidats")
plt.xlabel("Nombre d'années d'expérience")
plt.ylabel("Nombre de candidats")
plt.xticks(rotation=0)
plt.show()
```

```python
offres["description"].value_counts()
```

```python
offres["experience_extraite"] = (
    offres["description"]
    .str.extract(r"(\d+)", expand=False)
)

offres["experience_extraite"] = pd.to_numeric(
    offres["experience_extraite"],
    errors="coerce"
)

offres[["description", "experience_extraite"]].head(10)
```

```python
offres["description"].value_counts().sort_index().plot(
    kind="bar",
    figsize=(8, 5)
)

plt.title("Distribution de l'expérience professionnelle requise dans les offres")
plt.xlabel("Nombre d'années d'expérience requise")
plt.ylabel("Nombre d'offres")
plt.xticks(rotation=0)
plt.show()
```
